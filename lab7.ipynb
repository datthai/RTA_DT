```python
import pandas as pd
import numpy as np
import random
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split

# Reproducibility
np.random.seed(42)
random.seed(42)

print("Libraries imported.")
```

```python
# Task 1: Load data from train.csv into a pandas DataFrame
df = pd.read_csv("train.csv")
df.head()
```

```python
# Task 2: Show number of rows and columns
print(f"Rows:    {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

# Bonus — see column names and dtypes
print("\nDataset info:")
df.info()
```

```python
# Task 3: Missing data handling — 3 strategies on 3 copies of the dataset

# First, see WHERE the missing data is
print("Missing values per column:")
print(df.isnull().sum())
print()

# We'll demonstrate each strategy on a COPY so the original df stays intact.

# --- Option 1: dropna() — remove rows with ANY missing value ---
# Best when: very few rows have missing data and we cannot guess the value.
# Worst when: many rows have missing data — we lose too much information.
# In Titanic: 'Cabin' has 687/891 missing → dropna() would delete 77% of the data.
# So this option is NOT good for Titanic as-is. We'd drop 'Cabin' first.
df_opt1 = df.dropna()
print(f"Option 1 (dropna):  {df_opt1.shape[0]} rows left "
      f"(lost {df.shape[0] - df_opt1.shape[0]} rows)")

# --- Option 2: drop columns with too many missing values ---
# Best when: a column is mostly empty and unlikely to help the model.
# In Titanic: 'Cabin' is 77% empty → drop it. Keep 'Age' and 'Embarked'.
df_opt2 = df.drop(columns=["Cabin"])
print(f"Option 2 (drop Cabin column):  {df_opt2.shape[1]} columns left")

# --- Option 3: fillna() with the mean — imputation ---
# Best when: a column is numeric, mostly present, and missing values are random.
# In Titanic: 'Age' has 177 missing values (numeric, 20%) → fill with mean is reasonable.
# 'Embarked' has only 2 missing → fill with the mode (most frequent value).
df_opt3 = df.copy()
df_opt3["Age"] = df_opt3["Age"].fillna(df_opt3["Age"].mean())
df_opt3["Embarked"] = df_opt3["Embarked"].fillna(df_opt3["Embarked"].mode()[0])
print(f"Option 3 (impute Age + Embarked):  missing now = {df_opt3.isnull().sum().sum() - df_opt3['Cabin'].isnull().sum()}")
```

```python
# WHICH COLUMNS DID I CHOOSE FOR EACH OPTION AND WHY?
#
# Option 1 (dropna) — would be applied to a clean version of the data, NOT the raw
#   Titanic. Raw data has 'Cabin' with 77% missing, so dropna would delete most rows.
#   In practice: drop 'Cabin' first, then dropna() on the rest only kills ~20% of rows.
#
# Option 2 (drop columns) — I chose 'Cabin' because 687 of 891 values are missing
#   (77%). With so little signal, imputing would invent data, and modeling on it
#   would be misleading. Better to remove the column.
#
# Option 3 (fillna mean) — I chose 'Age' because it is numeric, missing 20% of the
#   time, and the mean is a reasonable central estimate. For 'Embarked' I used the
#   mode (most common port) since it is a categorical variable — mean has no meaning.
#
# The combined strategy I'll use below: drop 'Cabin', impute 'Age' and 'Embarked'.
```

```python
# Task 4: Use nunique() to find columns unsuitable for modeling
print("Unique values per column:")
print(df.nunique())
print()

# Columns to remove and WHY:
#   - PassengerId: 891 unique values → just an ID, no predictive power
#   - Name:        891 unique values → free-text, every row different
#   - Ticket:      681 unique values → mostly unique strings, hard to use without parsing
#   - Cabin:       147 unique values + 77% missing → unreliable
#
# A column where nunique() ≈ len(df) acts like a row index — it cannot generalize.
# We keep: Pclass, Sex, Age, SibSp, Parch, Fare, Embarked (and target: Survived).

df_clean = df.drop(columns=["PassengerId", "Name", "Ticket", "Cabin"])

# Now handle missing values on what's left (combined strategy from Task 3)
df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].mean())
df_clean["Embarked"] = df_clean["Embarked"].fillna(df_clean["Embarked"].mode()[0])

print(f"After cleaning: {df_clean.shape[0]} rows × {df_clean.shape[1]} columns")
print(f"Remaining missing values: {df_clean.isnull().sum().sum()}")
df_clean.head()
```

```python
# Task 5: Convert categorical variables with LabelEncoder
# 'Sex' and 'Embarked' are categorical (strings) — models need numbers.

le_sex = LabelEncoder()
df_clean["Sex"] = le_sex.fit_transform(df_clean["Sex"])
# Now: female=0, male=1 (or similar — depends on alphabetical order)

le_emb = LabelEncoder()
df_clean["Embarked"] = le_emb.fit_transform(df_clean["Embarked"])
# Now: C=0, Q=1, S=2

print("Sex mapping:     ", dict(zip(le_sex.classes_, le_sex.transform(le_sex.classes_))))
print("Embarked mapping:", dict(zip(le_emb.classes_, le_emb.transform(le_emb.classes_))))

df_clean.head()
```

```python
# Task 6: MinMaxScaler — bring all numeric features to the [0, 1] range
# Why: features like Fare (0–512) and Age (0–80) are on different scales.
# Algorithms that use distances (KNN, neural nets) are biased toward large-range features.
# MinMax brings everything to [0, 1] so each feature gets fair weight.

# Apply to FLOAT-like numeric columns. Survived is the target — don't scale it.
cols_to_scale = ["Age", "Fare", "SibSp", "Parch", "Pclass", "Sex", "Embarked"]

scaler = MinMaxScaler()
df_clean[cols_to_scale] = scaler.fit_transform(df_clean[cols_to_scale])

print("After scaling — every feature should have min=0 and max=1:")
df_clean.describe().loc[["min", "max"]]
```

```python
# Task 7: Split into train (80%) and test (20%)

X = df_clean.drop(columns=["Survived"])   # all features
y = df_clean["Survived"]                  # target

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,        # 20% test
    random_state=44,      # reproducibility — same split every run
)

print(f"X_train: {X_train.shape}")
print(f"X_test:  {X_test.shape}")
print(f"y_train: {y_train.shape}")
print(f"y_test:  {y_test.shape}")
```

```python
# Task 8: Random classifier that returns 0 or 1

def classify(x):
    """
    Single-case classifier — receives one row of features, returns a prediction.
    Random baseline: guess 0 or 1 with equal probability.
    """
    return random.randint(0, 1)


def run(f_classify, x):
    """Apply the classifier to every row."""
    return list(map(f_classify, x))


def evaluate(predictions, actual):
    """Count correct answers; return accuracy summary."""
    correct = list(filter(
        lambda item: item[0] == item[1],
        list(zip(predictions, actual))
    ))
    return f"{len(correct)} correct answers from {len(actual)}. Accuracy ({len(correct)/len(actual)*100:.0f}%)"


# Evaluate the random classifier on the training set
print("Random classifier on TRAIN set:")
print(evaluate(run(classify, X_train.values), y_train.values))

# And on the test set
print("\nRandom classifier on TEST set:")
print(evaluate(run(classify, X_test.values), y_test.values))

# COMMENT:
# A random binary classifier should score around 50% accuracy on average
# (random.randint(0,1) is a fair coin flip).
# In Titanic, ~62% of passengers died (Survived=0), so always predicting 0
# would already give ~62% — better than random! This is called the
# "majority class baseline" and a real model must beat it to be useful.
```

```python
# Bonus: compare with a real model (Random Forest)
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier(random_state=42)
rfc.fit(X_train, y_train)

train_preds = rfc.predict(X_train).tolist()
test_preds  = rfc.predict(X_test).tolist()

print("Random Forest on TRAIN set:")
print(evaluate(train_preds, y_train.values))

print("\nRandom Forest on TEST set:")
print(evaluate(test_preds, y_test.values))

# You should see Random Forest reach ~80%+ on test — much better than 50% random.
```
